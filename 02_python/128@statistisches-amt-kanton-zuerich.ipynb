## Dataset
# **Leerstehende Wohnungen im Kanton Zürich**

## Description

Die Leerwohnungszählung erfasst das Verhältnis von Angebot und Nachfrage auf dem Wohnungsmarkt. Durchgeführt wird sie nach den Vorgaben des Bundesamts für Statistik (BFS). Erhoben werden die Wohnungsleerstände (inkl. Einfamilienhäuser) von den Gemeinden. Gemäss Definition gilt eine Wohnung als leer, wenn sie am Stichtag (1. Juni) unbewohnt und bezugsbereit ist sowie  entweder zur dauernden Miete oder zum Kauf angeboten wird. Die Gemeinden können zur Erhebung der Wohnungsleerstände auf unterschiedliche Methoden zurückgreifen und diese bei Bedarf anpassen. Sie können zum Beispiel Registerdaten oder Daten der Elektrizitätswerke nutzen und bei Immobilienverwaltungen anfragen. In der Regel nutzen sie eine Kombination unterschiedlicher Erhebungsmethoden. Die Leerwohnungsziffer  (auch Leerwohnungsquote genannt) weist den Anteil der leeren Wohnungen am gesamten Wohnungsbestand aus. Aufgrund der freien Methodenwahl und dem Einfluss von Neubauprojekten ist die Vergleichbarkeit der Leerstände und -ziffern auf Gemeindeebene eingeschränkt.

```python
# Read a CSV dataset, trying comma and semicolon separators.
def get_dataset(url):
    if url[-3:] != "csv":
        print("The dataset URL does not end in '.csv'. Check that it points to a CSV file.\nPlease check the dataset link and adjust pandas' read_csv() parameters accordingly.")
    data = pd.read_csv(url, sep=",", on_bad_lines='warn', encoding_errors='ignore', low_memory=False)
    # If the DataFrame has one column or fewer, try ";" instead
    if data.shape[1] <= 1:
        data = pd.read_csv(url, sep=';', on_bad_lines='warn', encoding_errors='ignore', low_memory=False)
        if data.shape[1] <= 1:
            print("The data still has one column or fewer. The separator may be incorrect.\nPlease check the dataset manually and adjust the code.")
    return data
```
