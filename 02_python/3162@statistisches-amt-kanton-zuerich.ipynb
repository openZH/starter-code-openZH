# **Häusliche Gewalt im Kanton Zürich**

## Description

Das Monitoring häuslicher Gewalt im Kanton Zürich misst, wie verbreitet körperliche, seelische, sexuelle oder wirtschaftliche Gewalt in Familie und Partnerschaft ist. Die Kennzahlen stammen aus verschiedenen Quellen: Interventionsstelle gegen häusliche Gewalt, Obergericht Kanton Zürich, mannebüro Zürich, Team72, Bewährungs- und Vollzugsdienste, Kantonale Opferhilfe, Aufsuchender Dienst Forensic Nurses, Schutzunterkünfte für Minderjährige, Frauenhäuser des Kanton Zürich, Statthalterämter des Kanton Zürich und Staatsanwaltschaft. Jede Stelle erfasst nur jene Fälle, die bei ihr gemeldet und bearbeitet werden. Deshalb sind die Kennwerte, beispielsweise die Opferzahlen, nicht immer gleich. 
Die Daten werden jährlich aktualisiert und bilden die Grundlage der Webseite «Häusliche Gewalt in Zahlen» (siehe Landingpage). Dort finden Sie auch ein readme zu den Daten, inkl. Methodik und Limitationen.

```python
# Column types, completeness, and distinct non-missing values.
profile = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "missing": df.isna().sum(),
    "missing_pct": df.isna().mean().mul(100),
    "unique": df.nunique(dropna=True),
}).sort_values("missing_pct", ascending=False)
display(profile.round(1))
```

```python
# Show up to 20 columns with missing values; percentages use all rows.
missing = profile.loc[profile["missing"] > 0].head(20).rename_axis("column").reset_index()
if not missing.empty:
    fig, ax = plt.subplots(figsize=(8, max(2, len(missing) * 0.3)))
    sns.barplot(data=missing, x="missing_pct", y="column", color="steelblue", errorbar=None, ax=ax)
    ax.set(xlim=(0, 100), xlabel="Missing values (%)", ylabel="", title="Missingness by column")
    fig.tight_layout()
    plt.show()
else:
    print("No missing values to plot (or the dataset has no rows).")
```

```python
# Summarize numeric and non-numeric columns, including pandas string types.
if not df.empty:
    display(df.describe(include="all").T)
else:
    print("No data to summarize.")
```

```python
# Plot up to four numeric columns. Exclude missing/infinite values only from plots.
numeric = df.select_dtypes(include="number")
for column in numeric.columns[:4]:
    values = numeric[column].dropna()
    values = values.loc[~values.isin([float("inf"), -float("inf")])]
    if not values.empty:
        fig, ax = plt.subplots(figsize=(7, 3))
        sns.histplot(x=values, bins=30, ax=ax)
        ax.set(title=column, xlabel=column, ylabel="Count")
        fig.tight_layout()
        plt.show()
```
