## Description

Die Personalkennzahlen enthalten detaillierte Informationen zu den Angestellten der Direktionen und der Staatskanzlei, die dem Regierungsrat unterstellt sind. Die Behörden, die Rechtspflege sowie die selbstständigen und unselbstständigen Anstalten sind nicht Bestandteil der Personalkennzahlen. Die Kennzahlen stammen aus den jährlich veröffentlichten Personal- und Lohnstatistik und den Personalführungskennzahlen. Der Erhebungskreis kann je nach Quelle leicht abweichen; die massgebende Quelle ist bei jeder Ressource angegeben.
Es werden Anstellungen, Beschäftigungsumfang, Löhne und weitere Kennzahlen ausgewiesen. Diese werden unter anderem nach Merkmalen wie Geschlecht, Alter, Beschäftigungsgrad, Lohnklasse und Organisationseinheit differenziert.
Die Anzahl Anstellungen werden jeweils per Stichtag 31. Dezember des Berichtjahres erhoben, wobei jede Anstellung unabhängig ihres Beschäftigungsgrades voll gezählt wird. Die Anzahl Anstellungen ist höher als die Gesamtzahl der angestellten Personen, da eine Person mehrere Stellen innehaben kann (Mehrfachanstellung). Ausgewiesen werden dabei nur die Anstellungen der Kategorie Angestellte; die Kategorie übriges Personal (Praktikantinnen und Praktikanten, Lernende sowie Kommissionsmitglieder) wird in dieser Zählung nicht berücksichtigt.
Der Beschäftigungsumfang einer Verwaltungseinheit umfasst die Gesamtzahl aller während eines Jahres Beschäftigten in Vollzeiteinheiten. Eine Vollzeiteinheit entspricht einer Anstellung vom 1. Januar bis 31. Dezember bei einem Beschäftigungsgrad von 100 Prozent. Aus einer Anstellung im Teilzeit-Pensum bzw. durch unterjährigen Ein- bzw. Austritt ergibt sich der entsprechende Anteil einer Vollzeiteinheit. Der Beschäftigungsgrad wird aufgeteilt in Teilzeit: Beschäftigungsgrad < 50%, Teilzeit: Beschäftigungsgrad ≥ 50% und < 90% und Vollzeit: Beschäftigungsgrad ≥ 90%. 
Die Personalgruppe „Lehrpersonal“ enthält sämtliche Lehrpersonen der Volks-, Mittel- und Berufsschulen, ausgenommen Kurzkursleitende. Lehrpersonal findet sich in den Direktionen der Justiz und des Innern (Justizvollzug und Wiedereingliederung), der Bildungsdirektion (Volks-, Berufs- und Mittelschulen) und der Baudirektion (Amt für Natur und Landschaft).
Die verwendeten Begriffe sind im Glossar erläutert. Der Link dazu findet sich im Abschnitt 'Weitere Informationen'.

```python
def get_dataset(
    url: str, *, sep: str | None = None, drop_empty_unnamed: bool = True, **kwargs: object
) -> pd.DataFrame:
    """Read CSV data and remove fully empty columns with pandas-generated names."""
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
