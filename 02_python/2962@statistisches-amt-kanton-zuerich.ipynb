## Imports and helper functions

```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.style.use('ggplot')

params = {
    'text.color': (0.25, 0.25, 0.25),
    'figure.figsize': [18, 6],
   }

plt.rcParams.update(params)

import pandas as pd
```

```python
# Read a CSV dataset, trying comma and semicolon separators.
def get_dataset(url):
    if url[-3:] != "csv":
        print("The dataset URL does not end in '.csv'. Check that it points to a CSV file.\nPlease check the dataset link and adjust pandas' read_csv() parameters accordingly.")
    data = pd.read_csv(url, sep=",", on_bad_lines='warn', encoding_errors='ignore', low_memory=False)
    # If the DataFrame has one column or fewer, try ";" instead
    if data.shape[1] <= 1:
        data = pd.read_csv(url, sep=';', on_bad_lines='warn', encoding_errors='ignore', low_memory=False)
        if data.shape[1] <= 1:
            print("The data still has one column or fewer. The separator may be incorrect.\nPlease check the dataset manually and adjust the code.")
    return data
```
