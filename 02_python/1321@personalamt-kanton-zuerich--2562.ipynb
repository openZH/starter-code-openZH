## Description

Die Personalkennzahlen enthalten detaillierte Informationen zu den Angestellten der Direktionen und der Staatskanzlei, die dem Regierungsrat unterstellt sind. Die Behörden, die Rechtspflege sowie die selbstständigen und unselbstständigen Anstalten sind nicht Bestandteil der Personalkennzahlen. Die Kennzahlen stammen aus den jährlich veröffentlichten Personal- und Lohnstatistik und den Personalführungskennzahlen. Der Erhebungskreis kann je nach Quelle leicht abweichen; die massgebende Quelle ist bei jeder Ressource angegeben.<br>Es werden Anstellungen, Beschäftigungsumfang, Löhne und weitere Kennzahlen ausgewiesen. Diese werden unter anderem nach Merkmalen wie Geschlecht, Alter, Beschäftigungsgrad, Lohnklasse und Organisationseinheit differenziert.<br>Die Anzahl Anstellungen werden jeweils per Stichtag 31. Dezember des Berichtjahres erhoben, wobei jede Anstellung unabhängig ihres Beschäftigungsgrades voll gezählt wird. Die Anzahl Anstellungen ist höher als die Gesamtzahl der angestellten Personen, da eine Person mehrere Stellen innehaben kann &#40;Mehrfachanstellung&#41;. Ausgewiesen werden dabei nur die Anstellungen der Kategorie Angestellte; die Kategorie übriges Personal &#40;Praktikantinnen und Praktikanten, Lernende sowie Kommissionsmitglieder&#41; wird in dieser Zählung nicht berücksichtigt.<br>Der Beschäftigungsumfang einer Verwaltungseinheit umfasst die Gesamtzahl aller während eines Jahres Beschäftigten in Vollzeiteinheiten. Eine Vollzeiteinheit entspricht einer Anstellung vom 1. Januar bis 31. Dezember bei einem Beschäftigungsgrad von 100 Prozent. Aus einer Anstellung im Teilzeit-Pensum bzw. durch unterjährigen Ein- bzw. Austritt ergibt sich der entsprechende Anteil einer Vollzeiteinheit. Der Beschäftigungsgrad wird aufgeteilt in Teilzeit: Beschäftigungsgrad &lt; 50%, Teilzeit: Beschäftigungsgrad ≥ 50% und &lt; 90% und Vollzeit: Beschäftigungsgrad ≥ 90%. <br>Die Personalgruppe „Lehrpersonal“ enthält sämtliche Lehrpersonen der Volks-, Mittel- und Berufsschulen, ausgenommen Kurzkursleitende. Lehrpersonal findet sich in den Direktionen der Justiz und des Innern &#40;Justizvollzug und Wiedereingliederung&#41;, der Bildungsdirektion &#40;Volks-, Berufs- und Mittelschulen&#41; und der Baudirektion &#40;Amt für Natur und Landschaft&#41;.<br>Die verwendeten Begriffe sind im Glossar erläutert. Der Link dazu findet sich im Abschnitt &quot;Weitere Informationen&quot;.

This template targets pandas (3.x), seaborn (0.13+), matplotlib (3.x), and requests (2.x); packages are not installed or pinned by the notebook. Check the versions below when sharing results.


```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")
print(f"pandas {pd.__version__}; seaborn {sns.__version__}")
```

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpenZH-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
